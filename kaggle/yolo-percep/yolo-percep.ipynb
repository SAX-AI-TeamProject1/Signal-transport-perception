```python
import yaml

# data.yaml was generated on a local Mac, so its `path` points at a local absolute path —
# rewrite it to the Kaggle input mount before training.
DATASET_DIR = "/kaggle/input/yolo-gazebo-capture"
DATA_YAML = "/kaggle/working/data.yaml"

with open(f"{DATASET_DIR}/data.yaml") as f:
    cfg = yaml.safe_load(f)
cfg["path"] = DATASET_DIR
with open(DATA_YAML, "w") as f:
    yaml.safe_dump(cfg, f, allow_unicode=True, sort_keys=False)
print(cfg)
```

```python
from ultralytics import YOLO
import torch

MODEL = "yolov8n.pt"
EPOCHS = 50
IMG_SIZE = 960
BATCH = 8

def get_device() -> str:
    if torch.cuda.is_available():
        return "cuda"
    if torch.backends.mps.is_available():
        return "mps"
    return "cpu"

model = YOLO(MODEL)
model.train(
    data=DATA_YAML,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    batch=BATCH,
    device=get_device(),
    project="/kaggle/working/runs",
)
```
